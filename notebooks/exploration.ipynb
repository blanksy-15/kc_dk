# KCDK DraftKings Import Demo
Load the mock contest through the package and inspect the first weekly standings.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))

from kcdk.dk_import import import_contest, weekly_standings
from kcdk.members import load_members
```

```python
contest = import_contest(ROOT / 'data' / 'mock' / 'contest.csv')
members = load_members(ROOT / 'data' / 'members.csv')
standings = weekly_standings(contest, members)
```
